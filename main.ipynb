```python
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx                       # https://networkx.org/documentation/stable/reference/introduction.html
```

```python
# edges are bidirectional
#       - order of edge pairs does not matter
#       - therefore the graph is undirected
# there are NOT multiple edges between each pair of nodes

# best fit class: Graph()
G = nx.Graph()

G.add_edge(1, 2, de = 8, pe = 0.96, ce = 2)  # L1
G.add_edge(1, 3, de = 5, pe = 0.82, ce = 3)  # L2
G.add_edge(2, 3, de = 4, pe = 0.88, ce = 2)  # L3
G.add_edge(2, 4, de = 7, pe = 0.93, ce = 2)  # L4
G.add_edge(3, 4, de = 5, pe = 0.86, ce = 2)  # L5
G.add_edge(3, 5, de = 9, pe = 0.91, ce = 2)  # L6
G.add_edge(4, 5, de = 4, pe = 0.95, ce = 2)  # L7
G.add_edge(4, 6, de = 8, pe = 0.90, ce = 2)  # L8
G.add_edge(5, 6, de = 6, pe = 0.94, ce = 2)  # L9
```
